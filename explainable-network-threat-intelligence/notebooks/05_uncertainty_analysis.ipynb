```python
import numpy as np
import pandas as pd
import joblib
```

```python
model = joblib.load("../models/network_model.joblib")

X_test = joblib.load("../models/X_test.joblib")
y_test = joblib.load("../models/y_test.joblib")

feature_columns = joblib.load("../models/feature_columns.joblib")

X_test = X_test[feature_columns]

print("Test data shape:", X_test.shape)
```

```python
probabilities = model.predict_proba(X_test)

print(probabilities[:5])
```

```python
confidence = np.max(probabilities, axis=1)

uncertainty = 1 - confidence

print("Confidence:", confidence[:5])
print("Uncertainty:", uncertainty[:5])
```

```python
status = []

for value in confidence:
    if value < 0.70:
        status.append("Low Confidence")
    else:
        status.append("High Confidence")
```

```python
predictions = model.predict(X_test)

results = pd.DataFrame({
    "Actual": y_test,
    "Predicted": predictions,
    "Confidence": confidence,
    "Uncertainty": uncertainty,
    "Status": status
})

results.head(10)
```

```python
print(results["Status"].value_counts())
```

```python
results.to_csv("../models/uncertainty_results.csv", index=False)

print("Uncertainty analysis completed successfully!")
```
