## Phase 1 Summary

The AI4I 2020 Predictive Maintenance Dataset contains 10,000 records and 14 columns.

Important columns include:

- UDI
- Product ID
- Type
- Air temperature [K]
- Process temperature [K]
- Rotational speed [rpm]
- Torque [Nm]
- Tool wear [min]
- Machine failure

The target variable for our project is:

**Machine failure**

Where:

- 0 = No machine failure
- 1 = Machine failure

The dataset also contains five failure-mode columns:

- TWF - Tool Wear Failure
- HDF - Heat Dissipation Failure
- PWF - Power Failure
- OSF - Overstrain Failure
- RNF - Random Failure

In Phase 1, we inspected the dataset structure, data types, missing values, duplicate records, statistical information, machine types, and machine failure distribution.

The dataset is now ready for Exploratory Data Analysis (EDA).
